2-Création des première features (Features engeneering)

```python
# Copie de l'index temporel
features = pd.DataFrame({
    "frame_id": dataset["frame_id"]
})
```

```python
# 1. Visibilité du ballon
features["ball_visible"] = (
    dataset["x_b"].notna() &
    dataset["y_b"].notna()
).astype(int)
```

```python
# 2. Statistiques collectives par équipe
teams = {
    "home": range(1, 25),
    "away": range(25, 47)
}

for team_name, players in teams.items():

    visible_players = []
    player_speeds = []

    for i in players:

        x_col = f"x_{i}"
        y_col = f"y_{i}"
        s_col = f"s_{i}"

        # Un joueur est considéré visible si ses deux coordonnées existent
        visible = (
            dataset[x_col].notna() &
            dataset[y_col].notna()
        )

        visible_players.append(visible)

        # Vitesse uniquement lorsque le joueur est visible
        speed = dataset[s_col].where(visible)

        player_speeds.append(speed)

    # Nombre de joueurs visibles par frame
    visible_matrix = pd.concat(visible_players, axis=1)

    features[f"{team_name}_players_visible"] = (
        visible_matrix.sum(axis=1)
    )

    # Matrice des vitesses
    speed_matrix = pd.concat(player_speeds, axis=1)

    # Moyenne et maximum des vitesses disponibles
    features[f"{team_name}_mean_speed"] = (
        speed_matrix.mean(axis=1)
    )

    features[f"{team_name}_max_speed"] = (
        speed_matrix.max(axis=1)
    )
```

```python
# 3. Vérification de la table obtenue
print("Dimensions de la table :", features.shape)

print("\nPremières statistiques :")
display(features.describe().round(3))

print("\nValeurs manquantes :")
display(features.isna().sum())

print("\nVérification de l'alignement :")
print(
    "Même nombre de lignes :",
    len(features) == len(dataset)
)

print(
    "Même ordre de frame :",
    features["frame_id"].equals(dataset["frame_id"])
)
```

```python
# Dimensions du terrain selon la documentation
FIELD_LENGTH = 105.3
FIELD_WIDTH = 68.0

# Coordonnées du ballon
ball_x = dataset["x_b"].to_numpy()
ball_y = dataset["y_b"].to_numpy()

ball_valid = np.isfinite(ball_x) & np.isfinite(ball_y)

# Équipes et numéros de trajectoires
teams = {
    "home": range(1, 25),
    "away": range(25, 47)
}

for team_name, players in teams.items():

    # Récupération des coordonnées des joueurs
    x_players = dataset[
        [f"x_{i}" for i in players]
    ].to_numpy()

    y_players = dataset[
        [f"y_{i}" for i in players]
    ].to_numpy()

    # Joueur visible si ses deux coordonnées existent
    players_valid = (
        np.isfinite(x_players) &
        np.isfinite(y_players)
    )

    # Calcul des distances en mètres
    dx = (x_players - ball_x[:, None]) * FIELD_LENGTH
    dy = (y_players - ball_y[:, None]) * FIELD_WIDTH

    distances = np.sqrt(dx**2 + dy**2)

    # Une distance est valide seulement si le ballon
    # et le joueur sont tous les deux visibles
    valid_distances = players_valid & ball_valid[:, None]

    distances[~valid_distances] = np.nan

    # Distance minimale ballon-joueur
    safe_distances = np.where(
        valid_distances,
        distances,
        np.inf
    )

    min_distance = safe_distances.min(axis=1)

    min_distance[~np.isfinite(min_distance)] = np.nan

    features[f"{team_name}_nearest_ball_dist"] = min_distance

    # Nombre de joueurs à proximité du ballon
    for radius in [2, 5, 10]:

        features[f"{team_name}_players_within_{radius}m"] = (
            (valid_distances & (distances <= radius)).sum(axis=1)
        )


# Différence entre les distances minimales
features["nearest_dist_diff"] = (
    features["home_nearest_ball_dist"] -
    features["away_nearest_ball_dist"]
)


# Vérification des résultats
new_columns = [
    "home_nearest_ball_dist",
    "away_nearest_ball_dist",
    "nearest_dist_diff",
    "home_players_within_2m",
    "home_players_within_5m",
    "home_players_within_10m",
    "away_players_within_2m",
    "away_players_within_5m",
    "away_players_within_10m"
]

print("Dimensions de features :", features.shape)

print("\nStatistiques des nouvelles features :")
display(features[new_columns].describe().round(3))

print("\nValeurs manquantes :")
display(features[new_columns].isna().sum())

print("\nVérification de l'alignement :")
print(features["frame_id"].equals(dataset["frame_id"]))
```

```python
# Identifier les frames où le ballon est visible
ball_visible_mask = features["ball_visible"].eq(1)

# Les nombres de joueurs proches sont inconnus
# lorsque le ballon est invisible
proximity_columns = [
    f"{team}_players_within_{radius}m"
    for team in ["home", "away"]
    for radius in [2, 5, 10]
]

features.loc[
    ~ball_visible_mask,
    proximity_columns
] = np.nan

# Vérification des distances et des proximités
print("=== FRAMES AVEC BALLON VISIBLE ===")

print(
    "Nombre de frames :",
    ball_visible_mask.sum()
)

print("\n=== STATISTIQUES CONDITIONNELLES ===")

display(
    features.loc[ball_visible_mask, [
        "home_nearest_ball_dist",
        "away_nearest_ball_dist",
        "nearest_dist_diff",
        "home_players_within_2m",
        "home_players_within_5m",
        "home_players_within_10m",
        "away_players_within_2m",
        "away_players_within_5m",
        "away_players_within_10m"
    ]].describe().round(3)
)

print("\n=== VALEURS MANQUANTES ===")

display(features[proximity_columns].isna().sum())

print("\n=== ALIGNEMENT ===")

print(
    features["frame_id"].equals(dataset["frame_id"])
)
```

```python
# 1. Identifier l'équipe la plus proche du ballon
dist_diff = features["nearest_dist_diff"]

features["home_closer"] = np.where(
    dist_diff.isna(),
    np.nan,
    np.where(
        dist_diff < 0, 1,
        np.where(dist_diff > 0, 0, 0.5)
    )
)

# 2. Variation de l'avantage de proximité
# On calcule uniquement entre deux frames consécutives
dist_change = dist_diff.diff()

consecutive_frames = (
    features["frame_id"].diff() == 1
)

features["nearest_dist_diff_change"] = (
    dist_change.where(consecutive_frames)
)

# 3. Proportion des 5 dernières frames exploitables
# 1 = Barcelona plus proche
# 0 = Athletic Club plus proche
# 0.5 = égalité
features["home_closer_ratio_5"] = (
    features["home_closer"]
    .rolling(window=5, min_periods=3)
    .mean()
)

# Vérification
new_temporal_columns = [
    "home_closer",
    "nearest_dist_diff_change",
    "home_closer_ratio_5"
]

print("Dimensions de features :", features.shape)

print("\nStatistiques des nouvelles features :")
display(
    features[new_temporal_columns]
    .describe()
    .round(3)
)

print("\nValeurs manquantes :")
display(
    features[new_temporal_columns].isna().sum()
)

print("\nVérification de l'alignement :")
print(
    features["frame_id"].equals(dataset["frame_id"])
)
```

```python
# Récupérer les variations calculées
changes = features["nearest_dist_diff_change"].dropna()

print("=== DISTRIBUTION DES VARIATIONS ===")

print("Nombre de variations calculables :", len(changes))

print("\nQuantiles :")
display(
    changes.abs().quantile(
        [0.50, 0.90, 0.95, 0.99, 0.995, 0.999, 1.00]
    ).to_frame("variation_absolue_metres")
)

# Fréquence des grandes variations
print("\n=== FREQUENCE DES GRANDES VARIATIONS ===")

for threshold in [1, 2, 5, 10, 20]:
    count = (changes.abs() > threshold).sum()

    print(
        f"Variation > {threshold} m : "
        f"{count} frames ({count / len(changes) * 100:.3f} %)"
    )

# Examiner les 10 variations les plus importantes
print("\n=== 10 PLUS GRANDES VARIATIONS ===")

extreme_indices = (
    features["nearest_dist_diff_change"]
    .abs()
    .nlargest(10)
    .index
)

diagnostic = features.loc[
    extreme_indices,
    [
        "frame_id",
        "ball_visible",
        "nearest_dist_diff",
        "nearest_dist_diff_change",
        "home_nearest_ball_dist",
        "away_nearest_ball_dist"
    ]
].copy()

display(
    diagnostic.sort_values(
        "nearest_dist_diff_change",
        key=lambda x: x.abs(),
        ascending=False
    ).round(3)
)

# Vérifier l'alignement
print("\nAlignement :", features["frame_id"].equals(dataset["frame_id"]))
```

```python
pred = features["baseline_prediction"]

# Nombre de changements de classe consécutifs
changes = pred.ne(pred.shift())

print("=== NOMBRE DE CHANGEMENTS DE CLASSE ===")
print("Total :", changes.sum() - 1)

# Longueur des séquences consécutives
groups = changes.cumsum()

spells = (
    features.assign(spell_id=groups)
    .groupby(["spell_id", "baseline_prediction"])
    .size()
    .reset_index(name="duration_frames")
)

print("\n=== DURÉE DES SÉQUENCES PAR CLASSE ===")
display(
    spells.groupby("baseline_prediction")["duration_frames"]
    .describe()
    .round(2)
)

print("\n=== 20 SÉQUENCES LES PLUS LONGUES ===")
display(
    spells.sort_values("duration_frames", ascending=False)
    .head(20)
    .reset_index(drop=True)
)

print("\n=== SÉQUENCES TRÈS COURTES (1 À 3 FRAMES) ===")
display(
    spells[spells["duration_frames"] <= 3]
    ["baseline_prediction"]
    .value_counts()
    .to_frame("nombre_sequences")
)
```

```python
# ==========================================
# DIAGNOSTIC DES FRAMES X
# ==========================================

x_mask = features["baseline_prediction"].eq("X")

ball_visible = features["ball_visible"].eq(1)
dist_diff = features["nearest_dist_diff"]
ratio = features["home_closer_ratio_5"]

# Initialisation des motifs d'abstention
reasons = pd.Series("Autre cas", index=features.index, dtype="object")

reasons.loc[~ball_visible] = "Ballon invisible"

reasons.loc[
    ball_visible & dist_diff.isna()
] = "Distances indisponibles"

reasons.loc[
    ball_visible & dist_diff.notna() & ratio.isna()
] = "Historique temporel indisponible"

reasons.loc[
    ball_visible & dist_diff.notna() & ratio.notna() &
    dist_diff.abs().le(1)
] = "Écart de distance insuffisant"

reasons.loc[
    ball_visible & dist_diff.lt(-1) & ratio.lt(0.5)
] = "Désaccord avec la tendance temporelle"

reasons.loc[
    ball_visible & dist_diff.gt(1) & ratio.ge(0.5)
] = "Désaccord avec la tendance temporelle"

# On conserve uniquement les frames prédites X
x_reasons = reasons.loc[x_mask]

print("=== CAUSES DES PRÉDICTIONS X ===")

summary = x_reasons.value_counts().to_frame("nombre_frames")
summary["pourcentage_des_X"] = (
    summary["nombre_frames"] / x_mask.sum() * 100
).round(2)

display(summary)

# ==========================================
# ANALYSE DES SÉQUENCES X
# ==========================================

x_groups = (
    features.assign(
        is_x=x_mask,
        spell_id=x_mask.ne(x_mask.shift()).cumsum()
    )
    .loc[x_mask]
    .groupby("spell_id")
    .agg(
        debut=("frame_id", "min"),
        fin=("frame_id", "max"),
        duree=("frame_id", "size")
    )
    .reset_index()
)

print("\n=== STATISTIQUES DES SÉQUENCES X ===")
display(x_groups["duree"].describe().to_frame("duree_frames").round(2))

print("\n=== 15 PLUS LONGUES SÉQUENCES X ===")
display(
    x_groups.sort_values("duree", ascending=False)
    .head(15)
    .reset_index(drop=True)
)
```

```python
# ==========================================
# DIAGNOSTIC DES LONGUES SEQUENCES X
# ==========================================

# Identification des séquences consécutives
spell_id = (
    features["baseline_prediction"]
    .ne(features["baseline_prediction"].shift())
    .cumsum()
)

temp = features[["frame_id", "baseline_prediction"]].copy()
temp["spell_id"] = spell_id

# Extraction des séquences X
x_groups = (
    temp[temp["baseline_prediction"] == "X"]
    .groupby("spell_id")
    .agg(
        debut=("frame_id", "min"),
        fin=("frame_id", "max"),
        duree=("frame_id", "size")
    )
    .reset_index()
)

# Les 15 séquences les plus longues
top_x = x_groups.nlargest(15, "duree").copy()

# Calcul des statistiques de chaque séquence
results = []

for _, spell in top_x.iterrows():

    mask = features["frame_id"].between(
        spell["debut"], spell["fin"]
    )

    data_part = dataset.loc[mask]
    feature_part = features.loc[mask]

    # Actions dominantes
    action_mode = data_part["action"].mode()
    team_mode = data_part["team_action"].mode()

    results.append({
        "spell_id": spell["spell_id"],
        "debut": spell["debut"],
        "fin": spell["fin"],
        "duree": spell["duree"],

        "ball_visible_pct": round(
            feature_part["ball_visible"].mean() * 100, 2
        ),

        "home_players_mean": round(
            feature_part["home_players_visible"].mean(), 2
        ),

        "away_players_mean": round(
            feature_part["away_players_visible"].mean(), 2
        ),

        "action_dominante": (
            action_mode.iloc[0] if len(action_mode) else None
        ),

        "team_action_dominante": (
            team_mode.iloc[0] if len(team_mode) else None
        )
    })

diagnostic_x = pd.DataFrame(results)

print("=== DIAGNOSTIC DES 15 PLUS LONGUES SEQUENCES X ===")
display(diagnostic_x)
```

```python
# ==========================================
# ANALYSE DES X AVEC BALLON VISIBLE
# ==========================================

# Frames X alors que le ballon est détecté
mask = (
    features["baseline_prediction"].eq("X") &
    features["ball_visible"].eq(1)
)

x_visible = features.loc[mask].copy()

print("=== VOLUME ===")
print("Frames X avec ballon visible :", len(x_visible))
print(
    "Pourcentage de toutes les frames :",
    round(len(x_visible) / len(features) * 100, 2),
    "%"
)

# ==========================================
# 1. DISTANCES AU BALLON
# ==========================================

print("\n=== STATISTIQUES DES DISTANCES ===")

display(
    x_visible[
        [
            "home_nearest_ball_dist",
            "away_nearest_ball_dist",
            "nearest_dist_diff"
        ]
    ].describe().round(2)
)

# ==========================================
# 2. NOMBRE DE JOUEURS PROCHES DU BALLON
# ==========================================

proximity_columns = [
    "home_players_within_2m",
    "away_players_within_2m",
    "home_players_within_5m",
    "away_players_within_5m",
    "home_players_within_10m",
    "away_players_within_10m"
]

print("\n=== JOUEURS PROCHES DU BALLON ===")
display(x_visible[proximity_columns].mean().round(2).to_frame("moyenne"))

# ==========================================
# 3. DISTRIBUTION DES MOTIFS D'ABSTENTION
# ==========================================

dist_diff = features["nearest_dist_diff"]
ratio = features["home_closer_ratio_5"]

reasons = pd.Series("Autre", index=features.index, dtype="object")

reasons.loc[
    features["ball_visible"].eq(1) & dist_diff.isna()
] = "Distances indisponibles"

reasons.loc[
    features["ball_visible"].eq(1) &
    dist_diff.notna() &
    ratio.isna()
] = "Historique indisponible"

reasons.loc[
    features["ball_visible"].eq(1) &
    dist_diff.notna() &
    dist_diff.abs().le(1)
] = "Écart inférieur ou égal à 1 m"

reasons.loc[
    features["ball_visible"].eq(1) &
    dist_diff.lt(-1) &
    ratio.lt(0.5)
] = "Désaccord tendance / Barcelone plus proche"

reasons.loc[
    features["ball_visible"].eq(1) &
    dist_diff.gt(1) &
    ratio.ge(0.5)
] = "Désaccord tendance / Athletic plus proche"

print("\n=== MOTIFS DES X AVEC BALLON VISIBLE ===")

summary = reasons.loc[mask].value_counts().to_frame("nombre_frames")
summary["pourcentage"] = (
    summary["nombre_frames"] / len(x_visible) * 100
).round(2)

display(summary)
```

```python
# ==========================================
# SIMULATION DE PLUSIEURS SEUILS
# ==========================================

thresholds = [0.5, 0.75, 1.0, 1.25]

dist_diff = features["nearest_dist_diff"]
ratio = features["home_closer_ratio_5"]
ball_visible = features["ball_visible"].eq(1)

results = []

for threshold in thresholds:

    home_condition = (
        ball_visible &
        dist_diff.lt(-threshold) &
        ratio.ge(0.5)
    )

    away_condition = (
        ball_visible &
        dist_diff.gt(threshold) &
        ratio.lt(0.5)
    )

    prediction = pd.Series("X", index=features.index)
    prediction.loc[home_condition] = "H"
    prediction.loc[away_condition] = "A"

    results.append({
        "seuil_metres": threshold,
        "frames_H": (prediction == "H").sum(),
        "frames_A": (prediction == "A").sum(),
        "frames_X": (prediction == "X").sum(),
        "couverture_pct": round(
            prediction.isin(["H", "A"]).mean() * 100, 2
        ),
        "frames_recuperees": (
            prediction.isin(["H", "A"]) &
            features["baseline_prediction"].eq("X")
        ).sum()
    })

comparison = pd.DataFrame(results)

print("=== COMPARAISON DES SEUILS ===")
display(comparison)
```

```python
# ==========================================
# COMPARAISON TEMPORELLE : 1.00 m vs 0.75 m
# ==========================================

# Baseline actuelle
prediction_100 = features["baseline_prediction"].copy()

# Nouvelle candidate à 0.75 m
threshold = 0.75

dist_diff = features["nearest_dist_diff"]
ratio = features["home_closer_ratio_5"]
ball_visible = features["ball_visible"].eq(1)

prediction_075 = pd.Series("X", index=features.index)

home_condition = (
    ball_visible &
    dist_diff.lt(-threshold) &
    ratio.gt(0.5)
)

away_condition = (
    ball_visible &
    dist_diff.gt(threshold) &
    ratio.lt(0.5)
)

prediction_075.loc[home_condition] = "H"
prediction_075.loc[away_condition] = "A"


# ==========================================
# FONCTION D'ANALYSE DES SÉQUENCES
# ==========================================

def analyze_temporal_prediction(prediction, version):

    changes = prediction.ne(prediction.shift()).sum() - 1

    spell_id = prediction.ne(prediction.shift()).cumsum()

    spells = pd.DataFrame({
        "spell_id": spell_id,
        "prediction": prediction
    }).groupby(
        ["spell_id", "prediction"]
    ).size().reset_index(name="duration")

    summary = []

    for classe in ["H", "A", "X"]:

        class_spells = spells.loc[
            spells["prediction"].eq(classe), "duration"
        ]

        summary.append({
            "version": version,
            "classe": classe,
            "nombre_sequences": len(class_spells),
            "duree_mediane": class_spells.median(),
            "duree_moyenne": class_spells.mean(),
            "sequences_1_3_frames": class_spells.le(3).sum()
        })

    return changes, pd.DataFrame(summary)


# ==========================================
# COMPARAISON DES DEUX VERSIONS
# ==========================================

changes_100, summary_100 = analyze_temporal_prediction(
    prediction_100, "Seuil 1.00 m"
)

changes_075, summary_075 = analyze_temporal_prediction(
    prediction_075, "Seuil 0.75 m"
)

print("=== CHANGEMENTS DE CLASSE ===")

display(pd.DataFrame({
    "version": ["Seuil 1.00 m", "Seuil 0.75 m"],
    "nombre_changements": [changes_100, changes_075]
}))

print("\n=== DURÉE ET FRAGMENTATION DES SÉQUENCES ===")

display(
    pd.concat([summary_100, summary_075], ignore_index=True)
    .round(2)
)

print("\n=== COUVERTURE ===")

display(pd.DataFrame({
    "version": ["Seuil 1.00 m", "Seuil 0.75 m"],
    "couverture_pct": [
        prediction_100.isin(["H", "A"]).mean() * 100,
        prediction_075.isin(["H", "A"]).mean() * 100
    ]
}).round(2))
```

```python
# ==========================================
# COMPARAISON FRAME PAR FRAME
# ==========================================

comparison_frames = pd.DataFrame({
    "frame_id": features["frame_id"],
    "prediction_1m": prediction_100,
    "prediction_075m": prediction_075
})

# Frames dont la classe a changé
changed_frames = comparison_frames.loc[
    comparison_frames["prediction_1m"] !=
    comparison_frames["prediction_075m"]
].copy()

print("=== NOMBRE DE FRAMES MODIFIÉES ===")
print(len(changed_frames))

print("\n=== TRANSITIONS ENTRE LES CLASSES ===")
display(
    pd.crosstab(
        changed_frames["prediction_1m"],
        changed_frames["prediction_075m"],
        margins=True
    )
)

print("\n=== EXEMPLES DE FRAMES MODIFIÉES ===")
display(changed_frames.head(20))

print("\n=== VÉRIFICATION ===")
print(
    "Toutes les nouvelles prédictions sont H, A ou X :",
    prediction_075.isin(["H", "A", "X"]).all()
)
```

```python
# Création de prédictions avec des règles strictement symétriques

ball_visible = features["ball_visible"].eq(1)
dist_diff = features["nearest_dist_diff"]
ratio = features["home_closer_ratio_5"]

def make_prediction(threshold):
    prediction = pd.Series(
        "X",
        index=features.index,
        dtype="object"
    )

    home_condition = (
        ball_visible &
        dist_diff.lt(-threshold) &
        ratio.gt(0.5)
    )

    away_condition = (
        ball_visible &
        dist_diff.gt(threshold) &
        ratio.lt(0.5)
    )

    prediction.loc[home_condition] = "H"
    prediction.loc[away_condition] = "A"

    return prediction


# Les deux seuils utilisent exactement les mêmes règles
prediction_100 = make_prediction(1.0)
prediction_075 = make_prediction(0.75)

# On conserve les deux versions dans features
features["baseline_prediction_075"] = prediction_075

print("=== COMPARAISON DES DEUX SEUILS ===")
print(pd.crosstab(
    prediction_100,
    prediction_075,
    margins=True
))

print("\n=== DISTRIBUTION DU CANDIDAT 0.75 m ===")
print(prediction_075.value_counts())

print("\n=== COUVERTURE ===")
print(
    "1 m :",
    round(prediction_100.isin(["H", "A"]).mean() * 100, 2),
    "%"
)
print(
    "0.75 m :",
    round(prediction_075.isin(["H", "A"]).mean() * 100, 2),
    "%"
)

print("\n=== VALIDATION ===")
print(
    "Valeurs autorisées :",
    prediction_075.isin(["H", "A", "X"]).all()
)
print(
    "Alignement avec dataset :",
    features["frame_id"].equals(dataset["frame_id"])
)
```
